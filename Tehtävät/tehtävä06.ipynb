```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
centers = np.array([[221, 120], [218, 109], [209, 112], [215, 100], [222, 118], [212, 120]])
forwards = np.array([[200, 96], [202, 102], [210, 100], [194, 100], [211, 99], [218, 101]])
guards = np.array([[190, 88], [197, 93], [180, 75], [186, 90], [188, 87], [193, 90]])

print(centers.shape[0], forwards.shape[0], guards.shape[0])
```

```python
categories = ['centers', 'forwards', 'guards']
print(categories)
```

```python
players = np.stack((centers, forwards, guards))
print(players.shape)
```

```python
print(np.max(players[:, :, 0]))
print(np.max(players[:, :, 1]))
```

```python
for i in range(len(categories)):
    print(categories[i], np.mean(players[i, :, 0]))
```

```python
for i in range(len(categories)):
    plt.scatter(players[i, :, 0], players[i, :, 1], label=categories[i])
plt.xlabel('Height')
plt.ylabel('Weight')
plt.legend()
plt.show()
```

```python
plt.boxplot(players[:, :, 0].T, tick_labels=categories)
plt.ylabel('Height')
plt.show()
```
