```python
import numpy as np
```

```python
arvos = np.array([5, 5, 4, 5])
print(arvos)

opintop = np.array([5, 3, 5, 5])
print(opintop)
```

```python
print(np.average(arvos))
```

```python
print(np.sum(arvos * opintop) / np.sum(opintop))
print(np.average(arvos, weights=opintop))
print(np.matmul(arvos, opintop) / np.sum(opintop))
```
